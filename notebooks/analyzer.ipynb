```python
from core.videos import UrlVideoSource

source = await UrlVideoSource.new("https://api.apify.com/v2/key-value-stores/NmjTBbquqMg8SmoAC/records/video-catfunny36-20251003011608-7556797968859352334.mp4", "./videos")
source.get_video_file_path()
```

```python
from core.videos import VideoFile, FilesystemVideoSource, AudioFile

file = VideoFile(source)
print(f"duration: {file.get_duration()}, frames: {file.get_total_frames()}")
```

```python
from core.agents.video import ClipTaggerClient, FrameAnalyzer

ct_client = ClipTaggerClient()

analyzer = FrameAnalyzer(ct_client, file)
contents = await analyzer.get_n_frames()
contents
```
