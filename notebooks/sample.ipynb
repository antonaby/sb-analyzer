```python

from core.vloader import download_video

url = 'https://www.youtube.com/shorts/LX2sA07CX2M'

v_info = download_video(url, './videos')
video_file = v_info.get("requested_downloads", [])[0]["filepath"]
print(f"Final file: {video_file}")
```

```python
from core.vloader import extract_video

frames = extract_video(video_file)
```

```python
from openai import OpenAI
import os

client = OpenAI(
  base_url="https://api.inference.net/v1",
  api_key=os.getenv("INFERENCE_API_KEY")
)
```

```python
import pandas as pd

from core.analyzer import caption_single_frame

results = []
for i, frame in enumerate(frames):
  index, caption, base64_img = caption_single_frame(client, frame, i)
  results.append({
    'timestamp': frames[index]['timestamp'],
    'frame_number': frames[index]['frame_number'],
    **caption,
    'base64': base64_img
  })

pd.set_option("display.max_columns", None)

df = pd.DataFrame(results)
df
```
